# Moldova Orientation Check

Checks whether Moldova CXR images are mirror-reversed by showing images where
the radiologist annotated a cavity on **one side only**.

**Normal CXR convention:** patient faces you, so:
- Patient LEFT lung → appears on **RIGHT** side of screen
- Patient RIGHT lung → appears on **LEFT** side of screen

**Setup:** Only attach `mabdullahi454/tb-portals-cxr-pngs`. No other dataset needed.
The image lists are pre-computed from the local CSVs and hardcoded below.

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def load(fname, size=128):
    p = PNG_DIR / fname
    if not p.exists(): return None
    return np.array(Image.open(p).convert('L').resize((size, size)))

def plot_grid(png_names, title, out_path, cols=10, thumb=128):
    imgs = [(fn, load(fn, thumb)) for fn in png_names]
    imgs = [(fn, img) for fn, img in imgs if img is not None]
    if not imgs:
        print('No images found'); return
    rows = (len(imgs) + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(cols*1.5, rows*1.6))
    axes = np.array(axes).flatten()
    for ax, (fn, img) in zip(axes, imgs):
        ax.imshow(img, cmap='gray', aspect='auto')
        ax.set_xticks([]); ax.set_yticks([])
    for ax in axes[len(imgs):]:
        ax.axis('off')
    fig.suptitle(title, fontsize=10, y=1.01, color='darkred')
    plt.tight_layout()
    fig.savefig(out_path, dpi=100, bbox_inches='tight')
    plt.show()
    print(f'Saved {out_path} ({len(imgs)} images)')
```

```python
# == QUANTITATIVE CHECK: cardiac (heart) position, Moldova vs reference countries ==
# The heart is a bright radio-opaque shadow sitting on the patient's LEFT.
# In a NORMAL PA chest X-ray, patient-LEFT maps to IMAGE-RIGHT, so the lower-central
# region is brighter on the image-RIGHT half. If an image is MIRRORED, the heart
# shadow moves to the image-LEFT half. Unlike a cavity, the heart is in EVERY image,
# so this is a robust, objective, label-free orientation test.
#
# Prediction: Kazakhstan + Romania (shown normal by our stats) -> asym > 0 (heart image-right)
#             Moldova (suspected mirrored)                      -> asym < 0 (heart image-left)

import pandas as pd

MANIFEST = PNG_DIR.parent / 'manifest.csv'
if not MANIFEST.exists():
    hits = list(Path('/kaggle/input').rglob('manifest.csv'))
    MANIFEST = hits[0] if hits else MANIFEST
mf = pd.read_csv(MANIFEST)
mf['png'] = mf['image_path'].apply(lambda x: x.split('/')[-1])
print('manifest:', MANIFEST, '| countries:', mf['country'].value_counts().to_dict())

def sample_country(country, n=200, seed=0):
    sub = mf[mf['country'] == country]
    return sub.sample(min(n, len(sub)), random_state=seed)['png'].tolist()

def cardiac_asymmetry(fname, size=256):
    """>0 => heart on image-RIGHT (normal). <0 => heart on image-LEFT (mirrored)."""
    img = load(fname, size)
    if img is None:
        return None
    roi = img[int(0.40*size):int(0.72*size), :].astype(float)   # mid-cardiac band
    left  = roi[:, :size//2].mean()
    right = roi[:, size//2:].mean()
    return (right - left) / ((right + left) / 2 + 1e-6)          # exposure-normalised

def country_scores(country, n=200):
    s = [cardiac_asymmetry(f) for f in sample_country(country, n)]
    return np.array([x for x in s if x is not None])

print()
print(f'{"country":12s} {"n":>4s} {"mean_asym":>10s} {"median":>8s} {"%heart-image-RIGHT":>20s}  verdict')
print('-'*70)
results = {}
for c in ['Kazakhstan', 'Romania', 'Moldova']:
    s = country_scores(c)
    results[c] = s
    frac_right = float((s > 0).mean())
    verdict = 'NORMAL' if s.mean() > 0 else 'MIRRORED'
    print(f'{c:12s} {len(s):>4d} {s.mean():>+10.3f} {np.median(s):>+8.3f} '
          f'{frac_right*100:>18.0f}%  {verdict}')
print()
print('If Moldova\'s sign is OPPOSITE to Kazakhstan/Romania -> Moldova is mirror-reversed.')
```

```python
# == VISUAL PROOF: mean (average) CXR per country ==
# Averaging ~200 X-rays cancels random disease and leaves the consistent anatomy.
# The cardiac silhouette shows up as a bright blob displaced to ONE side.
# Normal country -> blob RIGHT of red line.  Mirrored -> blob LEFT of red line.

def mean_image(country, n=200, size=256):
    acc, c = np.zeros((size, size)), 0
    for f in sample_country(country, n):
        img = load(f, size)
        if img is not None:
            acc += img; c += 1
    return acc / max(c, 1), c

countries = ['Kazakhstan', 'Romania', 'Moldova']
fig, axes = plt.subplots(1, len(countries), figsize=(5*len(countries), 5.5))
for ax, c in zip(axes, countries):
    mi, nc = mean_image(c)
    ax.imshow(mi, cmap='gray')
    ax.axvline(x=128, color='red', ls='--', lw=1.2)
    side = 'RIGHT (normal)' if results[c].mean() > 0 else 'LEFT (MIRRORED)'
    ax.set_title(f'{c} — mean of {nc} CXRs\nheart blob on {side}', fontsize=10)
    ax.axis('off')
fig.suptitle('Mean CXR per country — heart (bright blob) position reveals orientation\n'
             'Kazakhstan/Romania = normal reference; Moldova should be the mirror image',
             fontsize=12)
plt.tight_layout()
fig.savefig(OUT_DIR / 'mean_cxr_by_country.png', dpi=120, bbox_inches='tight')
plt.show()
print('Saved: mean_cxr_by_country.png')
```

## How to read the result

### Quantitative test (the objective proof) — `cardiac asymmetry`
The heart is in **every** X-ray, so we don't need to find cavities. We measure
which side of centre the bright cardiac shadow sits on, across ~200 images per country:

- **Normal orientation:** heart on the **image-RIGHT** half → asymmetry score **> 0**
- **Mirrored:** heart on the **image-LEFT** half → asymmetry score **< 0**

Kazakhstan and Romania are our normal references (our stats already showed they
behave normally). **If Moldova's sign is the opposite of theirs, Moldova is mirrored.**
The `mean_cxr_by_country.png` figure makes this visible: the bright heart blob in the
Moldova average sits on the opposite side from Kazakhstan/Romania.

### Visual spot-check (corroborating) — the one-sided-cavity grids
The red dashed line is the centre of the image.
